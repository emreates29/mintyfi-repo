```python
from transformers import pipeline
import time
import pandas as pd

classifier = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")

```

```python
texts = [
    "I really like this new phone.",
    "The food at the restaurant was bad.",
    "Technology is changing the world."
]
```

```python
classification_results = []

for text in texts:
    start_time = time.time()
    result = classifier(text)
    end_time = time.time()
    
    classification_results.append({
        "Girdi": text,
        "Çıktı etiketi": result[0]['label'],
        "Olasılık": round(result[0]['score'], 4),
        "Süre": round(end_time - start_time, 4)
    })

pd.DataFrame(classification_results)
```

```python
df_classification = pd.DataFrame(classification_results)

with open("hf_results.md", "w", encoding="utf-8") as f:
    f.write("## Text Classification Sonuçları\n\n")
    f.write(df_classification.to_markdown(index=False))

print("hf_results.md")
```
