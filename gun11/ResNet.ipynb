```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image

model_resnet = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
model_resnet.eval()

transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

img = Image.open("ornek.jpg")
x = transform(img).unsqueeze(0)  

with torch.no_grad():
    output_resnet = model_resnet(x)

print("ResNet18 çıkış boyutu:", output_resnet.shape)  
```

```python
from transformers import AutoTokenizer, AutoModel

tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
model_bert = AutoModel.from_pretrained("distilbert-base-uncased")

text = "Transfer learning makes deep learning faster."
inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    outputs_bert = model_bert(**inputs)

print("DistilBERT last_hidden_state boyutu:", outputs_bert.last_hidden_state.shape)  
```
