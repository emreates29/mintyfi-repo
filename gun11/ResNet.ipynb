```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model_resnet = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
model_resnet = model_resnet.to(device)
model_resnet.eval()

transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

img = Image.open("ornek.jpg").convert("RGB")
x = transform(img).unsqueeze(0).to(device)

features = torch.nn.Sequential(*list(model_resnet.children())[:-1])

with torch.no_grad():
    embedding = features(x)
    output = model_resnet(x)  

embedding = embedding.view(embedding.size(0), -1)  

print("ResNet18 embedding boyutu:", embedding.shape)
print("ResNet18 çıktı boyutu:", output.shape)
```
